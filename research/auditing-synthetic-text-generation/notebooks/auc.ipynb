```python
jobs = JobList.from_urls(job_urls.values())
```

```python
def get_metrics(job: Job):
    estimates = job.get_node("estimate_privacy")
    metrics = estimates.get_metrics()
    fpr = np.array(metrics["trade_off_curves"]["fpr"])
    tpr = 1 - np.array(metrics["trade_off_curves"]["roc"])
    auc = np.trapz(tpr, fpr)
    return {"fpr": fpr, "tpr": tpr, "auc": auc}
get_metrics(jobs[0])
```

```python
metrics = [get_metrics(job) for job in jobs]
```

```python
results = pd.DataFrame({
    "run": job_urls.keys(),
    "auc": [m["auc"] for m in metrics],
})
results
```
