```python
import numpy as np
from privacy_estimates.experiments.aml import JobList
from matplotlib import pyplot as plt
```

```python
jobs = JobList.from_urls({
    "in_fraction=0.75, n_ref_models=8": "https://ml.azure.com/runs/careful_lychee_d34g9fx57b?wsid=/subscriptions/acc09744-1ee3-4242-b375-93421c63af0c/resourcegroups/PPML/workspaces/M365Research-PPML-EUS&tid=72f988bf-86f1-41af-91ab-2d7cd011db47#",
})
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 5))
for job in jobs:
    metrics = job.get_node("estimate_privacy").get_metrics()
    fpr = metrics["trade_off_curves"]["fpr"]
    roc_tpr = 1 - np.array(metrics["trade_off_curves"]["roc"])
    ax[0].plot(fpr, roc_tpr, label=job.local_name)
    ax[1].plot(fpr, roc_tpr, label=job.local_name)
ax[0].plot([0, 1], [0, 1], "--", color="gray")
ax[1].plot([0, 1], [0, 1], "--", color="gray")
ax[0].legend(loc="lower right")
ax[0].set_xlabel("False Positive Rate")
ax[0].set_ylabel("True Positive Rate")
ax[1].legend(loc="lower right")
ax[1].set_xlabel("True Positive Rate")
ax[1].set_ylabel("False Positive Rate")
ax[1].set_xscale("log")
ax[1].set_yscale("log")
plt.show()

```
